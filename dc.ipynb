```python
def get_characters(page_name):
    url = f"https://dc.fandom.com/api.php?action=parse&page={page_name}&format=json"
    response = requests.get(url)
    data = response.json()
    page_html = data['parse']['text']['*']
    soup = BeautifulSoup(page_html, 'html.parser')
    characters = {}
    for b_tag in soup.find_all('b'):
        if b_tag.text in target:
            parent_p = b_tag.find_parent('p')
            char_list = parent_p.find_next_sibling('ul')
            names = [(a.text, a.get('href')) for a in char_list.find_all('a')]
            characters[b_tag.text] = names

    return characters
```

```python
def get_portrait_url(page_name):
    url = f"https://dc.fandom.com/api.php?action=parse&page={page_name}&format=json"
    response = requests.get(url)
    data = response.json()
    page_html = data['parse']['text']['*']
    soup = BeautifulSoup(page_html, 'html.parser')
    infobox_img = soup.find('img', class_='pi-image-thumbnail')
    if infobox_img is not None:
        return infobox_img.get('src')
    else:
        return None
```

```python
def generate_poster(issue_characters_dict, categories_to_include, poster_size, title_text):
    poster_width, poster_height = poster_size
    poster_image = Image.new('RGBA', (poster_width, poster_height), (20,20,30,255))
    size_scales = [0.75, 0.55, 0.40, 0.30]
    layers = []

    for i, category in enumerate(categories_to_include):
        char_name, char_href = issue_characters_dict[category][0]
        clean_name = char_name.replace(' ', '_').replace('(', '').replace(')', '')
        filepath = f'images_nobg/{clean_name}_nobg.png'
        char_image = Image.open(filepath)
        orig_width, orig_height = char_image.size
        target_height = int(poster_height * size_scales[i])
        scale_factor = target_height / orig_height
        target_width = int(orig_width * scale_factor)
        char_resize = char_image.resize((target_width, target_height))

        if i == 0:
            pos_x = poster_width - target_width - 50 if i % 2 == 1 else 50
            pos_y = int(poster_height * 0.05)
        else:
            pos_x = poster_width - target_width - 150 if i % 2 == 1 else 50
            pos_y = poster_height - target_height - 200
        layers.append((char_resize, pos_x, pos_y))

    for char_image,x,y in reversed(layers):
        poster_image.paste(char_image, (x, y), char_image)

    draw = ImageDraw.Draw(poster_image)
    font = ImageFont.truetype('C:/Windows/Fonts/arialbd.ttf', 80)

    text_bbox = draw.textbbox((0,0), title_text, font=font, stroke_width= 5)
    text_width = text_bbox[2] - text_bbox[0]
    text_x = (poster_width - text_width) // 2
    text_y = poster_height - 150

    draw.text((text_x, text_y), title_text, font=font,stroke_width= 5,stroke_fill=(0,0,0,255), fill = (255, 255, 255, 255))
    return poster_image
```

```python
def create_gradient_color(width, height, top_color, bottom_color):
    gradient = np.zeros((height, width, 3), dtype = np.uint8)
    for y in range(height):
        ratio = y / height
        r = int(top_color[0] * (1 - ratio) + bottom_color[0] * ratio)
        g = int(top_color[1] * (1 - ratio) + bottom_color[1] * ratio)
        b = int(top_color[2] * (1 - ratio) + bottom_color[2] * ratio)
        gradient[y, :] = [r, g ,b]
    gradient_img = Image.fromarray(gradient).convert('RGBA')
    return gradient_img
```
